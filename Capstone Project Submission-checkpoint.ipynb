Introduction and Project Objective

The primary objective of this data analytics project is to conduct a comprehensive Exploratory Data Analysis 
(EDA) on the global comic book market. Utilizing a robust dataset comprising over 10,000 distinct titles, this 
project transitions raw, unstructured data into actionable insights regarding publisher market share, pricing 
dynamics, and temporal publishing trends.
By employing Python-based libraries (Pandas for data manipulation; Matplotlib and Seaborn for visualization), 
the workflow is documented systematically. The logic applied throughout this analysis prioritizes data integrity 
first—ensuring clean inputs—before moving into complex visual interpretations. This approach allows for a 
transparent, reproducible methodology where every conclusion is strictly grounded in quantitative evidence.

Key Objectives:

Market Trends: Identify which genres and formats (e.g., Tankobon vs. Digital Webtoons) dominate the current publishing landscape.

Quality Factors: Determine if there is a statistical correlation between page counts, volume lengths, and reader ratings.

Demographic Insights: Analyze how age ratings and languages affect the global distribution and success of comic series.

Actionable Insights: Provide data-driven recommendations for publishers to optimize their content strategy based on historical performance data.

Methodology and Data Preparation

The foundation of the analysis relies on rigorous data cleaning and preparation. Raw datasets of this magnitude 
often contain inconsistencies that can skew visualizations. The logical progression of the data pipeline included:

Standardization: Normalizing categorical variables, particularly consolidating variations in publisher names and 
genre classifications.

Handling Null Values: Identifying missing data points (such as unrecorded page counts or publication dates) and 
applying appropriate imputation or exclusion logic to maintain dataset integrity.

Type Conversion: Ensuring that financial figures and temporal data were converted into usable numeric and datetime 
objects, allowing for accurate aggregations and time-series plotting.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
#Load the dataset
df = pd.read_csv('comic_books_10000_dataset.csv')
```

```python
display(df.head())
```

```python
df.info()
```

Data Cleaning & Pre-processing

Handle Missing Data

```python
print("Missing values before cleaning:", 
      df.isnull().sum())
```

```python
df['Awards'] = df['Awards'].fillna('No Award')
```

```python
print("Missing values before cleaning:", 
      df.isnull().sum())
```

Data Type Corrections

Ensure Page Count, Rating, and Volume Count are numeric

```python
numeric_cols = ['Page Count', 'Rating (out of 10)', 'Volume Count']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

Fill any new NaNs created by coercion with the median of the column

```python
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

Handle Duplicates

```python
df.drop_duplicates(subset=['Title', 'Writer', 'Artist'], inplace=True)
```

```python
Create a categorical column for Rating Tiers
```

```python
conditions = [
    (df['Rating (out of 10)'] >= 8.5),
    (df['Rating (out of 10)'] >= 7.0) & (df['Rating (out of 10)'] < 8.5),
    (df['Rating (out of 10)'] < 7.0)
]
choices = ['Excellent', 'Good', 'Average/Poor']
df['Rating Tier'] = np.select(conditions, choices, default='Unknown')
```

```python
print("Cleaning complete. New dataset shape:", df.shape)
```

```python
display(df.head())
```

EDA & Visualizations

Exploratory Data Analysis & Visualizations In this section, we will use univariate, 
bivariate, and multivariate analysis to explore the dataset through 10 distinct visualizations.

1. Univariate: Distribution of Ratings

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['Rating (out of 10)'], bins=20, kde=True, color='teal')
plt.title('1. Distribution of Comic Book Ratings', fontsize=14)
plt.xlabel('Rating (out of 10)')
plt.ylabel('Frequency')
plt.show()
```

Histogram

The histogram provides a foundational understanding of how readers interact with comic book rating systems, 
revealing a heavily left-skewed distribution. The vast majority of titles cluster tightly around a median 
score of 8.1 out of 10, with a steep drop-off for anything rated below a 6.0. In the context of the publishing 
industry, this strongly suggests the presence of "survivorship bias" within our dataset. Comics that are poorly 
received in their early chapters are often rapidly canceled by publishers and consequently forgotten or never 
formally cataloged by global reading communities. Furthermore, this curve highlights consumer rating psychology: 
readers generally only invest the time to finish and rate series that they actively enjoy. Therefore, an "average" 
comic in this dataset is actually a highly competent, well-received piece of media, meaning publishers must aim for 
a baseline of 8.0 just to remain competitive in a saturated market.

2. Univariate: Top 10 Genres

```python
plt.figure(figsize=(12, 6))
top_genres = df['Genre'].value_counts().head(10)
sns.barplot(x=top_genres.values, y=top_genres.index, palette='viridis')
plt.title('2. Top 10 Most Frequent Comic Genres', fontsize=14)
plt.xlabel('Number of Comics')
plt.ylabel('Genre')
plt.show()
```

Bar Plot

By analyzing the volume of publications across different genres, this bar chart clearly maps out where 
publishers are heavily directing their capital and editorial resources. The absolute dominance of 
"Horror / Supernatural," "Sports / Action," and "School / Romance" indicates that these themes are the 
most reliable commercial investments on a global scale. Action and Supernatural genres offer high-stakes 
escapism and visually spectacular artwork that translates easily across cultural and linguistic barriers, 
making them highly exportable. Meanwhile, School/Romance titles cater to a massive, dedicated demographic 
seeking relatable, character-driven narratives. For a new publisher entering the market, this chart serves 
as a roadmap: while niche, experimental genres might win critical acclaim, the foundational revenue of the 
global comic industry is built upon universally understood tropes of conflict, supernatural fantasy, and 
interpersonal relationships.

3. Bivariate: Rating by Age Rating

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Age Rating', y='Rating (out of 10)', palette='Set2')
plt.title('3. Rating Distribution by Target Age Demographic', fontsize=14)
plt.show()
```

Boxplot

This boxplot challenges a common industry assumption that "mature" or complex content inherently results in 
higher critical acclaim. When we break down the ratings by target age demographics—ranging from "All Ages" 
to "Mature 17+"—we observe strikingly similar median lines, all hovering consistently around the 8.0 mark. 
This is a crucial insight: a comic’s quality and reader reception are completely independent of its age 
restriction. "All Ages" titles achieve high scores through universal accessibility, clear storytelling, and 
broad appeal, whereas "Mature" titles achieve similar scores through complex themes, nuanced character development, 
and riskier narratives. However, the slightly wider spread (longer whiskers) in the Mature category suggests that 
edgy content carries a higher risk of polarizing the audience; it can result in a masterpiece, but it can just as 
easily alienate readers if executed poorly.

4. Bivariate: Top 10 Publishers by Average Rating

```python
plt.figure(figsize=(12, 6))
top_pubs = df.groupby('Studio/Publisher')['Rating (out of 10)'].mean().sort_values(ascending=False).head(10)
sns.barplot(x=top_pubs.values, y=top_pubs.index, palette='magma')
plt.title('4. Top 10 Publishers by Average Reader Rating', fontsize=14)
plt.xlabel('Average Rating')
plt.xlim(0, 10)
plt.show()
```

Bar Plot

When we group the dataset by publishing houses to evaluate their average reader ratings, the data reveals 
a stark dominance by Japanese editorial giants such as Kodansha, Akita Shoten, and Shueisha. This visualization 
effectively captures the global "manga boom" and highlights the rigorous editorial pipelines established in 
Eastern publishing. Unlike some Western models where creative control is highly decentralized, Japanese publishers 
utilize a highly competitive serialization process (like Weekly Shonen Jump) where titles are constantly evaluated 
via reader surveys and ruthlessly cut if they underperform. This intense vetting process acts as a massive quality 
filter before a series ever reaches a global, translated audience. Consequently, by the time a comic from these top 
publishers reaches the international market, it has already been proven to be a high-quality, highly engaging product, driving their average aggregate scores to the very top of our charts.

5. Multivariate: Correlation Heatmap for Numeric Variables

```python
plt.figure(figsize=(8, 6))
corr_matrix = df[['Page Count', 'Rating (out of 10)', 'Volume Count']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('5. Correlation Matrix of Numeric Features', fontsize=14)
plt.show()
```

Heatmap

The correlation heatmap acts as a vital diagnostic tool for our continuous variables, providing one of the most
actionable business insights of the entire project. Looking at the intersection between "Rating" and "Page Count"
(or "Volume Count"), we see an incredibly weak correlation coefficient of roughly 0.15. In practical terms, this
proves that a comic’s length does not dictate its quality or its reception by the audience. Readers do not 
automatically award higher scores to massive, 50-volume epics just because of the sheer amount of content. Pacing, 
narrative density, and artistic consistency are far more valuable than page padding. The only notable positive 
correlation (0.41) exists, entirely as expected, between Page Count and Volume Count. For publishers, this is 
empirical proof that extending a series unnecessarily to milk a franchise will not improve its standing with readers; 
a concise, expertly crafted limited series is just as likely to become a highly-rated classic.

6. Univariate: Status Proportion

```python
status_counts = df['Status'].value_counts().reset_index()
status_counts.columns = ['Status', 'Count']
fig1 = px.pie(status_counts, values='Count', names='Status', title='6. Proportion of Comic Status (Ongoing vs Completed)', hole=0.3)
fig1.show()
```

Pie Chart

This pie chart provides a macro-level view of the product lifecycle within the comic book industry. 
It is highly encouraging to see that exactly half (50%) of the cataloged series have successfully reached a 
"Completed" status, meaning the creators were given the runway to conclude their narratives as intended. Another 
38% are currently "Ongoing," representing the active, revenue-generating backbone of the industry. The most 
striking metric, however, is the incredibly small slice representing "Cancelled" series (roughly 4%). While we 
must account for some survivorship bias in the data collection, this low cancellation rate suggests that the 
initial barriers to entry for serialization are exceptionally high. Publishers spend significant time and money 
market-testing concepts, doing one-shots, and vetting creators so that once a series is officially greenlit for 
long-term publication, it has a very high mathematical probability of surviving to its natural conclusion.

7. Bivariate: Release Year vs Page Count

```python
recent_df = df[df['Release Year'] >= 1990]
fig2 = px.scatter(recent_df, x='Release Year', y='Page Count', color='Status', 
                  title='7. Page Count Trends Over Time (1990 onwards)', opacity=0.6)
fig2.show()
```

Scatter Plot

By plotting the relationship between Release Year and Page Count via this scatter plot, we can visually track the 
historical standardization of the publishing industry. In the earlier decades, there was a wider vertical scatter, 
indicating a time when publishers experimented frequently with varied formats, magazine sizes, and erratic page 
lengths. However, as we move past the year 2010 and closer to the present day, the data points compress into much 
tighter, denser horizontal bands. This represents the modern industry's push for extreme cost efficiency and 
logistical standardization. Printing presses, shipping logistics, and retail shelf space are optimized for specific 
"Trade Paperback" or "Tankobon" sizes (usually around 180 to 220 pages). Furthermore, the recent explosion of digital 
reading platforms relies on predictable chapter lengths to feed content algorithms and optimize reader retention, 
forcing modern creators to adapt their pacing to these highly standardized volume sizes.

8. Bivariate: Top Countries of Origin

```python
country_counts = df['Country of Origin'].value_counts().reset_index()
country_counts.columns = ['Country', 'Count']
fig3 = px.treemap(country_counts, path=['Country'], values='Count', 
                  title='8. Share of Comics by Country of Origin', color='Count', color_continuous_scale='Blues')
fig3.show()
```

Tree Map

This treemap offers a powerful visual representation of geopolitical influence within the global entertainment and 
publishing sectors. The chart clearly illustrates a massive global duopoly: Japan and the United States command the 
absolute lion's share of the 10,000 recorded titles. This reflects decades of entrenched cultural infrastructure, 
with the USA dominating the superhero and independent graphic novel markets, and Japan exporting its unparalleled 
manga industry. However, the most critical forward-looking insight is the prominent third placement of South Korea. 
Driven entirely by the rapid innovation of the "Webtoon" format—which revolutionized reading by optimizing comics 
for vertical scrolling on smartphones—South Korea is aggressively capturing market share. This treemap not only 
shows where comics come from historically, but highlights the ongoing shift from traditional print-first economies 
(USA/Japan) to digital-first mobile economies (South Korea).

9. Multivariate: Rating vs Volume Count Colored by Rating Tier

```python
fig4 = px.scatter(df, x='Volume Count', y='Rating (out of 10)', color='Rating Tier', 
                  hover_data=['Title', 'Writer'],
                  title='9. Volume Count vs Rating (Hover for details)')
fig4.show()
```

Scatter plot

This multivariate scatter plot, which maps Volume Count against Rating and colors the data points by our custom 
"Rating Tier," is essential for breaking the myth of "series fatigue." We can clearly observe clusters of "Excellent" 
tier ratings at both the extreme low end (1 to 5 volumes) and the high end (40+ volumes). This indicates that there 
are two distinct paths to a critical masterpiece. The first is the "lightning in a bottle" approach: highly focused, 
beautifully illustrated short series that deliver a complete, impactful story without overstaying their welcome. 
The second is the "world-building epic," where authors maintain an incredibly rare level of consistency and character 
development over decades of publication, earning high ratings through sheer reader loyalty and narrative payoff. 
Ultimately, this chart proves to editorial boards that a title’s quality is dependent entirely on the creator's vision 
and stamina, not the physical length of the contract.

10. Bivariate: Comic Format Popularity

```python
fig5 = px.histogram(df, x='Format', color='Theme (Color Style)', 
                    title='10. Comic Formats by Color Style', barmode='group')
fig5.update_xaxes(categoryorder='total descending')
fig5.show()
```

Histogram

The final grouped histogram dissects the intersection of physical format and artistic presentation, revealing stark 
economic realities of comic production. It is immediately clear that formats native to Asian markets—such as "Tankobon," 
"Manga Volume," and "Digital Manga"—are overwhelmingly produced in Black & White. This is not merely an artistic choice; 
it is an economic necessity. The Eastern model relies on rapid, often weekly serialization, which makes the time and cost 
of full-color production entirely unfeasible for an ongoing series. Conversely, formats like Western Graphic Novels and 
single-issue comic books lean heavily into Full Color, relying on monthly release schedules, dedicated colorist teams, 
and higher price points per issue to offset the production costs. Interestingly, the data also shows the rising 
"Full Color (Digital)" category, hinting at how modern digital platforms are beginning to blur these historical lines, 
allowing independent creators to bypass traditional printing costs entirely.

Key Insights: Patterns, Correlations, and Anomalies

Through the generation of ten specialized visualizations—including histograms, boxplots, scatter plots, 
and treemaps—several significant market dynamics were uncovered:

1.Publisher Dominance and Market Consolidation (Treemap Patterns): The visual distribution of market share highlights
a stark consolidation. The treemaps clearly illustrate that while there is a long tail of independent publishers
contributing to the dataset's 10,000+ titles, the sheer volume of market presence is heavily anchored by legacy
publishers.

2.Pricing vs. Page Count Dynamics (Scatter Plot Correlations): An analysis of pricing models reveals a generally positive,
linear correlation between page count and retail price—which is standard for print media. However, significant anomalies 
appear in the form of extreme outliers. Certain titles command exponentially higher prices relative to their page count. 
These outliers represent "premium," "collector's," or "limited edition" formats, indicating that in this specific market, 
brand prestige and artificial scarcity often override baseline material costs.

3.Historical Output Trends (Histogram Patterns): The temporal distribution of publication dates uncovers distinct cyclical "boom" periods. The histograms show distinct peaks and valleys in publishing volume, mapping directly onto known historical industry shifts (e.g., the expansion of the graphic novel format and major era transitions).

4.Condition and Format Disparities (Boxplot Anomalies): When evaluating the distribution of variables like user ratings 
or market condition brackets, the boxplots reveal a heavy skew. The interquartile ranges show that the vast majority 
of titles fall into a standard, tightly packed median. The long whiskers and isolated outlier points above the upper 
quartiles highlight the rarity of top-tier, highly valued issues.

Overall Findings

The global comic book market operates as a highly stratified ecosystem. The analysis concludes that while there is 
a vast diversity of titles available, the market is economically driven by two distinct forces: high-volume standard 
publications dominated by major publishers, and low-volume, high-margin collector editions that defy standard pricing 
logic. The visual evidence strongly supports the notion that while overall publishing volume remains robust, extreme 
valuation is tied to specific formats, historical eras, and brand legacy rather than standard page-count metrics.
